This file is used run the knn regression in python and also record the running time.

```python
# import the needed packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time
from sklearn.neighbors import KNeighborsRegressor
```

```python
# define the parameters
trainNums=["1000","2000","4000","8000"]
testNums=["500","1000","2000"]
kValues=[1,3,5,7]
resultFilePath = "TestProject\\Data\\PythonResult.txt"
```

```python
resultFile=open(resultFilePath,"w")

for trainNum in trainNums:
    for testNum in testNums:
        for kValue in kValues:


            # read the data
            trainData="TestProject\\Data\\synthetic_train_"+trainNum+".txt"
            testData="TestProject\\Data\\synthetic_test_"+testNum+".txt"
            trainXy=pd.read_csv(trainData)
            testXy=pd.read_csv(testData)
            trainX=trainXy.iloc[:, :2].values
            trainy=np.squeeze(trainXy.iloc[:,2])
            testX=testXy.iloc[:, :2].values
            testy=np.squeeze(testXy.iloc[:,2])

            start_time = time.time()
            neigh = KNeighborsRegressor(n_neighbors=kValue)
            neigh.fit(trainX,trainy)
            predictResult=neigh.predict(testX)
            end_time = time.time()

            elapsed_time = end_time - start_time
            elapsed_time_ms=elapsed_time*1000

            resultFile.write(
                str(kValue)+","+trainNum+","+testNum+","+str(elapsed_time_ms)+'\n'
            )

            # write the result into the csv file
             
resultFile.close()


```
