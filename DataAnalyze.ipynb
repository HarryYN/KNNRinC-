```python
#import the packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


resultFilePath="TestProject\\Data\\result.txt"
resultData = pd.read_csv(resultFilePath)
```

```python
resultData.dtypes
```

```python
# draw the graphs for scalling
plt.scatter(resultData["numTrain"],resultData["time"])
plt.xlabel("Number of records in train dataset")
plt.ylabel("running time in ms")
plt.title("C# runtime experiment numTrain")
plt.show()

plt.scatter(resultData["numTest"],resultData["time"])
plt.xlabel("Number of records in test dataset")
plt.ylabel("running time in ms")
plt.title("C# runtime experiment numTest")
plt.show()

plt.scatter(resultData["k"],resultData["time"])
plt.xlabel("k value in knn")
plt.ylabel("running time in ms")
plt.title("C# runtime experiment k value")
plt.show()
```

From the code and graphs above, we can draw some simple conclusins.

1. The running time increases linearly with the number of training set data.

2. The running time increases linearly with the number of test set data. 

3. The k value has no impact on the final running time
